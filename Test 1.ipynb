1. Truncate strings to a max of 4 characters using in built function


```python
def truncate_string(string, length=4):
    return string[:length]

string = 'Hello I am Mohini'
print(truncate_string(string)) 
```

2. Convert each character to uppercase if it’s a vowel, else lowercase using inbuilt function


```python
def convert_character(string):
    vowels = 'aeiou'
    return ''.join(s.upper() if s.lower() in vowels else s.lower() for s in string)

string = "Keep Doctor Away"
print(convert_character(string))

```

3. Filter strings that do NOT end with a vowel using inbuilt functions


```python
def filter_strings(strings):
    vowels = ('a', 'e', 'i', 'o', 'u', 'A', 'E', 'I', 'O', 'U')
    return [s for s in strings if not s.endswith(vowels)]

strings = ["Hello", "I", "am", "Mohini"]
print(filter_strings(strings))

```

4. Convert a list of full names to initials using inbuilt functions


```python
def convert_initials(names):
    return [''.join(word[0].upper() for word in name.split()) for name in names]

names = ["Ananya pandey", "Shraddha kapoor", "Christopher briney"]
print(convert_initials(names))
```

5.Find Sum of Digits That Are NOT Multiples of 2 or 3 using recursion


```python
def not_multiple(n):
    if n == 0:
        return 0
    digit = n % 10
    if digit % 2 != 0 and digit % 3 != 0:
        return digit + not_multiple(n // 10)
    else:
        return not_multiple(n // 10)

number = 99999
print(not_multiple(number))
```

6.Count Total Prime Digits in a Number using recursion


```python
def is_prime(num):
    if num < 2:
        return False
    for i in range(2,int(num**0.5) + 1):
        if num % i == 0:
            return False
    return True
num = 90
print(is_prime(num))
```

7.Sum of Squares of Digits at Odd Positions using recursion


```python
def sum_squares_odd_positions(n, pos=1):
    if n == 0:
        return 0
    digit = n % 10
    if pos % 2 != 0: 
        return digit**2 + sum_squares_odd_positions(n // 10, pos + 1)
    else:
        return sum_squares_odd_positions(n // 10, pos + 1)

number = 98989
print(sum_squares_odd_positions(number))
```

8. check if number is digit-reducible to a single digit by summing using recursion

```python
def reducible(num):
    if num < 10:
        return True
    else:
        return reducible(sum(int(digit) for digit in str(num)))
num = 90
print(reducible(num))
```

9. for each string, double every vowel using in built function

```python
def double_vowel(string):
    vowels = 'aeiouAEIOU'
    return ''.join(s*2 if s in vowels else s for s in string)

string = 'I am right here'
print(double_vowel(string))
```

```python
def check_title_case(words):
    return [str(word.istitle()) for word in words]
words = ['i bhargavi','bhargavi i']
result = check_title_case(words)
print(result)
```

11. Write a function rectangle_diagonal(length, width), to calculate the diagonal  of the rectangle 


```python
import math
```

```python
def rectangle_diagonal(length, width):
    return math.sqrt(length**2 + width**2)
length = 10
width = 5
diagonal = rectangle_diagonal(length, width)
print('Diagonal = ',diagonal)
```

13. Perform Modular Exponentiation, Write a function mod_exp(a, b, m) that returns (a^b) % m efficiently.


```python
def mod_exp(a, b, m):
    result = 1
    a = a % m  
    while b > 0:
        if b % 2 == 1:
            result = (result * a) % m
        b = b >> 1  
        a = (a * a) % m
    return result


a = 2
b = 5
m = 8
print(mod_exp(a, b, m))  
```

14. find all 3 digit numbers satisfying abc + cba = 1333 write a function to print all such 3 digit combinations.

```python
def find_3_digits_numbers():
    solutions = []
    for a in range(1,10): # a can not be 0
        for b in range(0,10):
            for c in range(1,10): # c cannot be 0
                abc = 100 * a + 10 * b + c
                cba = 100 * c + 10 * b + a
                if abc + cba == 1333:
                    solutions.append((abc),(cba))
                    print(f'{abc} + {cba} = 1333')

    return solutions
result = find_3_digits_numbers()
print('Solutions found = ')
print(result)
```
